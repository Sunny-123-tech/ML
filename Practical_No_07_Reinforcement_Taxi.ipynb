```python
import gymnasium as gym
import random
import numpy as np
```

```python
env = gym.make("Taxi-v4")
```

```python
alpha = 0.5
gamma = 0.99
epsilon = 0.1
episodes = 1000
```

```python
# Q = np.zeros((env.observation_space, env.action_space))
```

```python
Q = np.zeros((500, 6))
```

```python
def epsilon_greedy (state):
    if random.random() <= epsilon:
        return env.action_space.sample()
    else:
        return np.argmax(Q[state])
```

```python
for episode in range(episodes):
    done = False
    state, _ = env.reset()

    action = epsilon_greedy(state)
    totaL_reward = 0
    epsiode_length = 0

    while not done:
        next_state, reward, terminated, truncated, _ = env.step(action)
        done = terminated or truncated
        next_action = epsilon_greedy(next_state)
        if done:
            Q[state, action] += alpha * (reward  - Q[state, action])
        else:
            Q[state, action] += alpha * (reward + gamma * (Q[next_state, next_action] - Q[state, action]))

        state = next_state
        action = next_action
        totaL_reward += reward
        epsiode_length += 1

    print(f"epsiode = {episode+1}/1000 -> reward = {totaL_reward} & epsiode_len = {epsiode_length}")
env.close()
```

```python
test_env = gym.make("Taxi-v4", render_mode = 'human')

for episode in range(10):

    state, _ = test_env.reset()
    done = False
    total_reward = 0

    while not done:

        action = np.argmax(Q[state])

        state, reward, terminated, truncated, _ = test_env.step(action)

        done = terminated or truncated
        total_reward += reward

    print(f"Test Episode {episode+1}: Reward = {total_reward}")

test_env.close()
```

```python
rewards = []    
rewards.append(total_reward)
print("Average Reward of Last 100 Episodes:",
      np.mean(rewards[-100:]))
```
